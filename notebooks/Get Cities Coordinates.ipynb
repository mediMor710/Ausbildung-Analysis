```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
from math import sqrt
```

```python
df = pd.read_csv('analys_result.csv', index_col=0)
```

```python
df['city'] = df['ausbildung place'].str.strip().str.title()
city_counts = df['city'].value_counts().reset_index()
city_counts.columns = ['city', 'count']
```

```python
geo_locator = Nominatim(user_agent='ausbildung_student_project')
geo_code = RateLimiter(geo_locator.geocode, min_delay_seconds=1)

def get_coords(city):
    location = geo_code(f"{city}, Germany")
    return pd.Series([location.latitude, location.longitude] if location else [None, None])
```

```python
coords = pd.DataFrame({'city': city_counts['city']})
coords[['latitude', 'longitude']] = coords['city'].apply(get_coords)
coords.to_csv('city_coordinates.csv', index=False)
```

```python
from sentence_transformers import SentenceTransformer
from bertopic import BERTopic
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer
import nltk
from nltk.corpus import stopwords
from sklearn.cluster import KMeans

german_stopwords = stopwords.words('german')

umap_model = UMAP(
    n_neighbors=15,
    n_components=5,
    min_dist=0.0,
    metric='cosine', 
    random_state=42
)

embedding_model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')
```

```python
docs = df['firma opinion'].astype(str).tolist()

embeddings = embedding_model.encode(docs, show_progress_bar=True)
```

```python
docs2 = df['beruf opinion'].astype(str).tolist()

embeddings2 = embedding_model.encode(docs2, show_progress_bar=True)
```

```python
from bertopic.representation import MaximalMarginalRelevance

custom_stopwords = german_stopwords + ['gut', 'immer', 'viele', 'ausbildung', 'gibt', 'firma', 'ausbildungsberuf']
new_vectorizer = CountVectorizer(stop_words=custom_stopwords, ngram_range=(1,2))
mmr = MaximalMarginalRelevance(diversity=0.5)
```

```python
from gensim.corpora import Dictionary
from gensim.models.coherencemodel import CoherenceModel

new_vectorizer = CountVectorizer(stop_words=custom_stopwords, ngram_range=(1,2))
analyzer = new_vectorizer.build_analyzer()
tokens = [analyzer(d) for d in docs]
dictionary = Dictionary(tokens)

results = {}
for k in [4, 6, 7, 8, 9, 10, 12]:
    model = BERTopic(embedding_model=embedding_model, umap_model=umap_model,
                     hdbscan_model=KMeans(n_clusters=k, random_state=42),
                     vectorizer_model=new_vectorizer, representation_model=mmr)
    model.fit_transform(docs, embeddings)
    topic_words = [[w for w, _ in model.get_topic(t)] for t in model.get_topics()]
    coherence_model = CoherenceModel(topics=topic_words, texts=tokens,
                                     dictionary=dictionary, coherence='c_v').get_coherence()
    results[k] = round(coherence_model, 3)

print(results)
```

```python
final_model = BERTopic(embedding_model=embedding_model, umap_model=umap_model,
                       hdbscan_model=KMeans(n_clusters=7,random_state=42),
                       vectorizer_model=new_vectorizer, representation_model=mmr)
topics, _ = final_model.fit_transform(docs, embeddings)
df['firma_topics'] = topics
```

```python
print(final_model.get_topic_info())
```

```python
id2topic = {
    0: 'Zusammenarbeit und Arbeitsatmosphäre',
    1: 'Das Lernen durch Teams und die Integration mit dem Ansprechpartner',
    2: 'Lernen durch Projekte',
    3: 'Die Betreuung der Mitarbeiter',
    4: 'Schnell lernen und die Verantwortung nehmen',
    5: 'Dualstudium mit internationalen Studenten',
    6: 'Zufriedenheit mit dem Geld und der Arbeit'
}
```

```python
from gensim.corpora import Dictionary
from gensim.models.coherencemodel import CoherenceModel

new_vectorizer2 = CountVectorizer(stop_words=custom_stopwords, ngram_range=(1,2))
analyzer2 = new_vectorizer2.build_analyzer()
tokens2 = [analyzer2(d) for d in docs2]
dictionary2 = Dictionary(tokens2)

results = {}
for k in [4, 6, 7, 8, 9, 10, 12]:
    model2 = BERTopic(embedding_model=embedding_model, umap_model=umap_model,
                      hdbscan_model=KMeans(n_clusters=k, random_state=42),
                      vectorizer_model=new_vectorizer2, representation_model=mmr)
    model2.fit_transform(docs2, embeddings2)
    topic_words2 = [[w for w, _  in model2.get_topic(t)] for t in model2.get_topics()]
    c_v = CoherenceModel(topics=topic_words2, texts=tokens2,
        dictionary=dictionary2, coherence='c_v').get_coherence()

    results[k] = round(c_v, 3)

print(results)
```

```python
final_model2 = BERTopic(embedding_model=embedding_model, umap_model=umap_model,
                        hdbscan_model=KMeans(n_clusters=4, random_state=42),
                        vectorizer_model=new_vectorizer2, representation_model=mmr)
topics2, _ = final_model2.fit_transform(docs2, embeddings2)
df['firma topics'] = topics
df['beruf topics'] = topics2
```

```python
df.to_csv('bewertungen_with_topic_modeling.csv', index=False)
```
